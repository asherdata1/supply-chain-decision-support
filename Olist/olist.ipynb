```python
import pandas as pd
```

# Inspection

```python
orders = pd.read_csv(
    "olist_ecommerce_datasets/olist_orders_dataset.csv"
)

items = pd.read_csv(
    "olist_ecommerce_datasets/olist_order_items_dataset.csv"
)

products = pd.read_csv(
    "olist_ecommerce_datasets/olist_products_dataset.csv"
)

sellers = pd.read_csv(
    "olist_ecommerce_datasets/olist_sellers_dataset.csv"
)

customers = pd.read_csv(
    "olist_ecommerce_datasets/olist_customers_dataset.csv"
)

payments = pd.read_csv(
    "olist_ecommerce_datasets/olist_order_payments_dataset.csv"
)

for name, df in {
    "orders": orders,
    "items": items,
    "products": products,
    "sellers": sellers,
    "customers": customers,
    "payments": payments,
}.items():
    print(f"\n{name.upper()}")
    print(df.shape)
    display(df.info())
```

```python
print("Orders")
print("Rows:", len(orders))
print("Unique order IDs:", orders["order_id"].nunique())
print("Unique customer IDs:", orders["customer_id"].nunique())

print("\nItems")
print("Rows:", len(items))
print("Unique order IDs:", items["order_id"].nunique())
print("Unique product IDs:", items["product_id"].nunique())
print("Unique seller IDs:", items["seller_id"].nunique())

print("\nProducts")
print("Rows:", len(products))
print("Unique product IDs:", products["product_id"].nunique())

print("\nSellers")
print("Rows:", len(sellers))
print("Unique seller IDs:", sellers["seller_id"].nunique())

print("\nCustomers")
print("Rows:", len(customers))
print("Unique customer IDs:", customers["customer_id"].nunique())
print("Unique customer unique IDs:", customers["customer_unique_id"].nunique())

print("\nPayments")
print("Rows:", len(payments))
print("Unique order IDs:", payments["order_id"].nunique())
```

```python
items_per_order = items.groupby("order_id").size()
sellers_per_order = items.groupby("order_id")["seller_id"].nunique()
products_per_order = items.groupby("order_id")["product_id"].nunique()
payments_per_order = payments.groupby("order_id").size()

print("Items per order")
display(items_per_order.describe())

print("\nSellers per order")
display(sellers_per_order.value_counts().sort_index())

print("\nProducts per order")
display(products_per_order.describe())

print("\nPayments per order")
display(payments_per_order.value_counts().sort_index())
```

```python
date_cols = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date",
]

orders[date_cols] = orders[date_cols].apply(pd.to_datetime)

items["shipping_limit_date"] = pd.to_datetime(
    items["shipping_limit_date"]
)
```

```python
orders["order_status"].value_counts()
```

```python
timestamp_cols = [
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date",
]

missing_by_status = (
    orders.groupby("order_status")[timestamp_cols]
    .agg(lambda x: x.isna().mean())
    .round(3)
)

display(missing_by_status)
```

```python
delivered = orders.loc[
    orders["order_status"] == "delivered"
].copy()

delivered["approval_hours"] = (
    delivered["order_approved_at"]
    - delivered["order_purchase_timestamp"]
).dt.total_seconds() / 3600

delivered["seller_to_carrier_days"] = (
    delivered["order_delivered_carrier_date"]
    - delivered["order_approved_at"]
).dt.total_seconds() / 86400

delivered["carrier_to_customer_days"] = (
    delivered["order_delivered_customer_date"]
    - delivered["order_delivered_carrier_date"]
).dt.total_seconds() / 86400

delivered["total_delivery_days"] = (
    delivered["order_delivered_customer_date"]
    - delivered["order_purchase_timestamp"]
).dt.total_seconds() / 86400

display(
    delivered[
        [
            "approval_hours",
            "seller_to_carrier_days",
            "carrier_to_customer_days",
            "total_delivery_days",
        ]
    ].describe()
)
```

```python
anomalies = pd.Series({
    "approval_before_purchase": (
        delivered["order_approved_at"]
        < delivered["order_purchase_timestamp"]
    ).sum(),

    "carrier_before_approval": (
        delivered["order_delivered_carrier_date"]
        < delivered["order_approved_at"]
    ).sum(),

    "delivery_before_carrier": (
        delivered["order_delivered_customer_date"]
        < delivered["order_delivered_carrier_date"]
    ).sum(),

    "delivery_before_purchase": (
        delivered["order_delivered_customer_date"]
        < delivered["order_purchase_timestamp"]
    ).sum(),
})

display(anomalies)
```

```python
olist_df = (
    items
    .merge(
        orders,
        on="order_id",
        how="left",
        validate="many_to_one",
    )
    .merge(
        products,
        on="product_id",
        how="left",
        validate="many_to_one",
    )
    .merge(
        sellers,
        on="seller_id",
        how="left",
        validate="many_to_one",
    )
    .merge(
        customers,
        on="customer_id",
        how="left",
        validate="many_to_one",
    )
)
```

```python
payment_summary = (
    payments.groupby("order_id")
    .agg(
        payment_value=("payment_value", "sum"),
        payment_installments=("payment_installments", "max"),
        payment_count=("payment_sequential", "count"),
        payment_type_count=("payment_type", "nunique"),
    )
    .reset_index()
)
```

```python
primary_payment = (
    payments.loc[
        payments.groupby("order_id")["payment_value"].idxmax(),
        ["order_id", "payment_type"],
    ]
    .rename(columns={"payment_type": "primary_payment_type"})
)

payment_summary = payment_summary.merge(
    primary_payment,
    on="order_id",
    how="left",
    validate="one_to_one",
)
```

```python
olist_df = olist_df.merge(
    payment_summary,
    on="order_id",
    how="left",
    validate="many_to_one",
)
```

```python
print("Shape:", olist_df.shape)
print("Unique orders:", olist_df["order_id"].nunique())
print("Unique sellers:", olist_df["seller_id"].nunique())
print("Unique products:", olist_df["product_id"].nunique())

olist_df.info()
```

### Export consolidated dataset

```python
olist_df.to_csv(
    "olist_ecommerce_datasets/olist_consolidated.csv",
    index=False,
)
```

# EDA

```python
fulfillment = olist_df.dropna(
    subset=[
        "order_delivered_carrier_date",
        "shipping_limit_date",
    ]
).copy()

fulfillment["missed_shipping_deadline"] = (
    fulfillment["order_delivered_carrier_date"]
    > fulfillment["shipping_limit_date"]
)

print("Eligible rows:", len(fulfillment))
display(
    fulfillment["missed_shipping_deadline"]
    .value_counts()
    .to_frame("count")
    .assign(
        percent=lambda x: (
            x["count"] / x["count"].sum() * 100
        ).round(2)
    )
)
```

```python
fulfillment["allowed_fulfillment_days"] = (
    fulfillment["shipping_limit_date"]
    - fulfillment["order_approved_at"]
).dt.total_seconds() / 86400

fulfillment["allowed_fulfillment_days"].describe()
```

```python
fulfillment["allowed_fulfillment_days"].round().value_counts().sort_index()
```

```python
delivered_fulfillment = fulfillment.loc[
    fulfillment["order_status"] == "delivered"
].copy()

delivered_fulfillment["late_customer_delivery"] = (
    delivered_fulfillment["order_delivered_customer_date"]
    > delivered_fulfillment["order_estimated_delivery_date"]
)

display(
    pd.crosstab(
        delivered_fulfillment["missed_shipping_deadline"],
        delivered_fulfillment["late_customer_delivery"],
        normalize="index",
    ).round(3)
)
```

```python
seller_performance = (
    fulfillment.groupby("seller_id")
    .agg(
        orders=("order_id", "nunique"),
        miss_rate=("missed_shipping_deadline", "mean"),
    )
)

display(
    seller_performance.loc[
        seller_performance["orders"] >= 20
    ]["miss_rate"].describe()
)
```

```python
fulfillment["purchase_month"] = (
    fulfillment["order_purchase_timestamp"]
    .dt.to_period("M")
)

monthly_miss_rate = (
    fulfillment.groupby("purchase_month")[
        "missed_shipping_deadline"
    ]
    .agg(["count", "mean"])
)

display(monthly_miss_rate)
```
